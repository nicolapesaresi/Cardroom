```python
from cardroom.briscola.game.env import BriscolaEnv
from cardroom.briscola.game.player import BriscolaPlayer
from cardroom.briscola.agents.random import RandomAgent
from cardroom.briscola.agents.bot import BotAgent
from cardroom.briscola.agents.donatello import DonatelloAgent
from cardroom.briscola.agents.evaluation.evaluate_endgame import EndgameEvaluator, EndgameSolver
```

Plot an endgame tree and check the agent's play

```python
solver = EndgameSolver()
endgame = solver.generate_endgame()
policy, results = solver.solve_endgame()
```

```python
solver.plot_tree(policy, results)
```

```python
evaluator = EndgameEvaluator(agent=DonatelloAgent(simulations=200))
evaluator.evaluate_endgame(endgame)
```

Evaluate the agent's endgame abilities over many games

```python
results, fig = evaluator.evaluate_agent(n_endgames = 10000)
```
